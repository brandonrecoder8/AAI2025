## Self-Reflection / Critique-and-Revise Prompt

You are an editor reviewing a summary for first-year college students.

Source passage:
"Remote work allows employees to perform job duties outside a traditional office, often
from home or another location with internet access. Supporters say it can reduce commuting
time, give workers more flexibility, and expand hiring beyond a company's local area.
However, remote work can also create challenges, including communication delays, isolation,
difficulty separating work and personal time, and unequal access to suitable workspaces or
reliable internet. Effective remote-work policies often include clear communication
expectations, regular check-ins, secure technology practices, and flexibility based on job
responsibilities."

Original summary:
"Remote work is when people work from home. It is good because people do not have to commute,
but it can also be hard. Companies should use technology and make rules."

Task:
First, critique the original summary using the criteria below. Then write a revised summary.

Evaluation criteria:
1. Accuracy: Include only claims supported by the source passage.
2. Completeness: Cover at least two benefits and two challenges of remote work.
3. Audience: Use plain language appropriate for first-year college students.
4. Length: Write 80–100 words.
5. Organization: Use exactly one paragraph with a clear main idea and logical flow.
6. Focus: Explain both the benefits and limitations, then mention what effective policies
   should include.
7. Style: Avoid vague phrases such as "it is good" or "make rules."
8. Formatting: Return exactly these two labeled sections:
   Critique:
   Revised Summary:

In the critique, identify at least three specific weaknesses in the original summary.
Do not add facts that do not appear in the source passage.

## AI Output

### Critique:
The original summary is generally accurate, but it is too brief and vague. It identifies
reduced commuting as a benefit but omits other important benefits from the source, including
greater flexibility for workers and the ability for companies to hire beyond their local area.
It says remote work “can also be hard,” but it does not explain specific challenges such as
communication delays, isolation, blurred boundaries between work and personal life, or unequal
access to reliable internet and suitable workspaces. The phrase “make rules” is also unclear
because it does not describe the recommended communication, check-in, security, and flexibility
practices.

### Revised Summary:
Remote work allows employees to complete their jobs outside a traditional office, often from
home. It can reduce commuting time, provide more flexibility, and help companies hire workers
from a wider geographic area. However, remote work may also create communication delays,
isolation, difficulty separating work from personal life, and unequal access to reliable
internet or suitable workspaces. Effective remote-work policies should establish clear
communication expectations, include regular check-ins, promote secure technology practices,
and remain flexible according to employees’ job responsibilities.
